```python
# IMPORTANT:
# This notebook is for visualization of the finalized results only.
# It does not retrain the model or perform a new evaluation.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# ============================================================
# FINAL DEVELOPMENT / CROSS-VALIDATION RESULTS
# Source: image-only-final(1).ipynb
# This notebook only visualizes finalized results.
# It does NOT retrain or reevaluate the model.
# ============================================================

cv_results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Balanced Accuracy",
        "Macro Precision",
        "Macro Recall",
        "Macro F1"
    ],
    "Image Only": [
        0.7431,
        0.6377,
        0.5970,
        0.6377,
        0.6031
    ],
    "Image + Age": [
        0.7244,
        0.6861,
        0.6362,
        0.6861,
        0.6008
    ]
})

cv_results
```

```python
# ============================================================
# DEVELOPMENT / CROSS-VALIDATION PERFORMANCE COMPARISON
# ============================================================

x = np.arange(len(cv_results["Metric"]))
width = 0.35

fig, ax = plt.subplots(figsize=(11, 6))

bars1 = ax.bar(
    x - width/2,
    cv_results["Image Only"],
    width,
    label="Image Only"
)

bars2 = ax.bar(
    x + width/2,
    cv_results["Image + Age"],
    width,
    label="Image + Age"
)

ax.set_title(
    "ResNet50 Development Performance: Image Only vs Image + Age",
    fontsize=14,
    fontweight="bold"
)

ax.set_ylabel("Score")
ax.set_xticks(x)
ax.set_xticklabels(cv_results["Metric"], rotation=20, ha="right")
ax.set_ylim(0, 0.85)

ax.legend()
ax.grid(axis="y", alpha=0.3)

# Add values above all bars
for bars in [bars1, bars2]:
    for bar in bars:
        value = bar.get_height()
        ax.text(
            bar.get_x() + bar.get_width() / 2,
            value + 0.015,
            f"{value:.4f}",
            ha="center",
            va="bottom",
            fontsize=9
        )

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# FINAL HOLDOUT EVALUATION RESULTS
# Source: image-only-final(1).ipynb
# Final configuration: ResNet50 Image Only
# ============================================================

holdout_metrics = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Balanced Accuracy",
        "Macro Precision",
        "Macro Recall",
        "Macro F1"
    ],
    "Score": [
        0.6948,
        0.6432,
        0.6191,
        0.6432,
        0.6206
    ]
})

holdout_metrics
```

```python
# ============================================================
# FINAL HOLDOUT PERFORMANCE
# ============================================================

fig, ax = plt.subplots(figsize=(10, 6))

bars = ax.bar(
    holdout_metrics["Metric"],
    holdout_metrics["Score"]
)

ax.set_title(
    "Final ResNet50 Image-Only Holdout Performance",
    fontsize=14,
    fontweight="bold"
)

ax.set_ylabel("Score")
ax.set_ylim(0, 0.8)

ax.tick_params(axis="x", rotation=20)
ax.grid(axis="y", alpha=0.3)

# Add values above all bars
for bar in bars:
    value = bar.get_height()
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        value + 0.015,
        f"{value:.4f}",
        ha="center",
        va="bottom",
        fontsize=10
    )

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# FINAL HOLDOUT CLASSIFICATION REPORT
# Source: image-only-final(1).ipynb
# ============================================================

classification_results = pd.DataFrame({
    "Class": [
        "Emmetropia",
        "Myopia",
        "Hyperopia"
    ],
    "Precision": [
        0.37,
        0.92,
        0.57
    ],
    "Recall": [
        0.37,
        0.74,
        0.82
    ],
    "F1-Score": [
        0.37,
        0.82,
        0.67
    ]
})

classification_results
```

```python
# ============================================================
# CLASS-SPECIFIC HOLDOUT PERFORMANCE
# ============================================================

x = np.arange(len(classification_results["Class"]))
width = 0.25

fig, ax = plt.subplots(figsize=(10, 6))

bars1 = ax.bar(
    x - width,
    classification_results["Precision"],
    width,
    label="Precision"
)

bars2 = ax.bar(
    x,
    classification_results["Recall"],
    width,
    label="Recall"
)

bars3 = ax.bar(
    x + width,
    classification_results["F1-Score"],
    width,
    label="F1-Score"
)

ax.set_title(
    "Final ResNet50 Image-Only Class-Specific Performance",
    fontsize=14,
    fontweight="bold"
)

ax.set_ylabel("Score")
ax.set_xticks(x)
ax.set_xticklabels(classification_results["Class"])
ax.set_ylim(0, 1.0)

ax.legend()
ax.grid(axis="y", alpha=0.3)

# Add values above all bars
for bars in [bars1, bars2, bars3]:
    for bar in bars:
        value = bar.get_height()
        ax.text(
            bar.get_x() + bar.get_width() / 2,
            value + 0.015,
            f"{value:.2f}",
            ha="center",
            va="bottom",
            fontsize=9
        )

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# FINAL HOLDOUT CONFUSION MATRIX
# Source: image-only-final(1).ipynb
# ============================================================

confusion_matrix = np.array([
    [10, 3, 14],
    [13, 65, 10],
    [4, 3, 32]
])

class_names = [
    "Emmetropia",
    "Myopia",
    "Hyperopia"
]
```

```python
# ============================================================
# CONFUSION MATRIX VISUALIZATION
# ============================================================

fig, ax = plt.subplots(figsize=(8, 7))

sns.heatmap(
    confusion_matrix,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names,
    cbar=False,
    ax=ax
)

ax.set_title(
    "Final ResNet50 Image-Only Confusion Matrix",
    fontsize=14,
    fontweight="bold"
)

ax.set_xlabel("Predicted Class")
ax.set_ylabel("Actual Class")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# DEVELOPMENT CROSS-VALIDATION VS FINAL HOLDOUT PERFORMANCE
# ============================================================
# Development results:
# ResNet50 Image Only, 10-fold Stratified Group CV
#
# Holdout results:
# Final ResNet50 Image Only model
# 15% untouched patient-level holdout set
#
# This notebook only visualizes finalized results.
# It does NOT retrain or reevaluate the model.
# ============================================================

comparison_results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Balanced Accuracy",
        "Macro Precision",
        "Macro Recall",
        "Macro F1"
    ],
    "Development CV": [
        0.7431,
        0.6377,
        0.5970,
        0.6377,
        0.6031
    ],
    "Final Holdout": [
        0.6948,
        0.6432,
        0.6191,
        0.6432,
        0.6206
    ]
})

comparison_results
```

```python
# ============================================================
# DEVELOPMENT CV VS FINAL HOLDOUT PERFORMANCE
# ============================================================

x = np.arange(len(comparison_results["Metric"]))
width = 0.35

fig, ax = plt.subplots(figsize=(11, 6))

ax.bar(
    x - width/2,
    comparison_results["Development CV"],
    width,
    label="Development Cross-Validation"
)

ax.bar(
    x + width/2,
    comparison_results["Final Holdout"],
    width,
    label="Final Holdout Test"
)

ax.set_title(
    "Development Cross-Validation vs Final Holdout Test Performance",
    fontsize=14,
    fontweight="bold"
)

ax.set_ylabel("Score")
ax.set_xticks(x)
ax.set_xticklabels(
    comparison_results["Metric"],
    rotation=20,
    ha="right"
)

ax.set_ylim(0, 0.85)

ax.legend()
ax.grid(axis="y", alpha=0.3)

# Add values above bars
for i in range(len(comparison_results)):
    ax.text(
        i - width/2,
        comparison_results["Development CV"][i] + 0.015,
        f"{comparison_results['Development CV'][i]:.4f}",
        ha="center",
        va="bottom",
        fontsize=9
    )

    ax.text(
        i + width/2,
        comparison_results["Final Holdout"][i] + 0.015,
        f"{comparison_results['Final Holdout'][i]:.4f}",
        ha="center",
        va="bottom",
        fontsize=9
    )

plt.tight_layout()
plt.show()
```
